```python
from datasets import load_dataset

# 加载 wikitext-103-raw-v1 数据集
dataset = load_dataset("wikitext", "wikitext-103-raw-v1")

print(dataset)
print("\nExample sample:")
print(dataset['train'][0])
```

```python
import pandas as pd

df = pd.DataFrame(dataset['train'])
df
```

```python
import pandas as pd
import random
import re

# df 已经是 DataFrame，包含 column = "text"

def is_english_sentence(text):
    """判断是否为正常英文句子：包含英文单词，且非乱码/标题结构"""
    if not isinstance(text, str):
        return False
    
    # 至少包含一些英文字符
    if not re.search(r'[A-Za-z]', text):
        return False
    
    # 过滤掉 wiki 标题，例如 "= heading ="
    if re.match(r'\s*=\s*[^=]+=\s*', text):
        return False

    return True

def word_count(text):
    """统计英文单词数量"""
    return len(re.findall(r"[A-Za-z]+", text))

# 过滤：英文句子 & 单词数 >= 20
filtered_df = df[df["text"].apply(lambda x: is_english_sentence(x) and word_count(x) >= 20)]

print(f"Total eligible sentences: {len(filtered_df)}")

# 随机采样 5000 条（如果不够就采样全部）
sample_size = min(5000, len(filtered_df))

sample_df = filtered_df.sample(sample_size, random_state=42)

# 查看前 5 条
sample_df
```

```python
sample_df.to_csv("./wikitext_sample_5000.csv", index=False)
```
